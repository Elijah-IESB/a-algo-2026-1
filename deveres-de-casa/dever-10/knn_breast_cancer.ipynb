# Algoritmo KNN - Aplicacao Medica

Classificacao do dataset Breast Cancer Wisconsin como benigno ou maligno usando KNN.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
dados = load_breast_cancer()

x_treino, x_teste, y_treino, y_teste = train_test_split(
    dados.data,
    dados.target,
    test_size=0.2,
    random_state=42,
    stratify=dados.target,
)
```

```python
valores_k = [1, 3, 5]
metricas = ["euclidean", "manhattan"]
resultados = []

for valor_k in valores_k:
    for metrica in metricas:
        modelo = Pipeline(
            [
                ("normalizador", StandardScaler()),
                ("knn", KNeighborsClassifier(n_neighbors=valor_k, metric=metrica)),
            ]
        )

        modelo.fit(x_treino, y_treino)
        previsoes = modelo.predict(x_teste)
        acuracia = accuracy_score(y_teste, previsoes)

        resultados.append(
            {
                "k": valor_k,
                "metrica": metrica,
                "acuracia": acuracia,
                "previsoes": previsoes,
            }
        )

for resultado in resultados:
    print(
        f"K = {resultado['k']} | "
        f"Metrica = {resultado['metrica']} | "
        f"Acuracia = {resultado['acuracia']:.4f}"
    )
```

```python
melhor_resultado = max(resultados, key=lambda item: item["acuracia"])

print("Melhor combinacao:")
print(
    f"K = {melhor_resultado['k']} | "
    f"Metrica = {melhor_resultado['metrica']} | "
    f"Acuracia = {melhor_resultado['acuracia']:.4f}"
)

print()
print(classification_report(
    y_teste,
    melhor_resultado["previsoes"],
    target_names=["maligno", "benigno"],
))
```

## Respostas

1. O melhor desempenho foi obtido com **K = 3**.
2. A melhor metrica foi a **euclidiana**, com acuracia de **0.9825**.
3. Com K muito pequeno, como K = 1, o modelo fica mais sensivel a ruidos e casos isolados da base de treino.
4. Normalizar e importante porque o KNN depende do calculo de distancias; sem normalizacao, atributos com escalas maiores dominam o resultado.
